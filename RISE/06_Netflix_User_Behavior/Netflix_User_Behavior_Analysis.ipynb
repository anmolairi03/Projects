```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("netflix_behavior.csv")
```

```python
print(df.head())
```

```python
genre_watch = df.groupby("Genre")["Watch_Hours"].sum().sort_values(ascending=False)
plt.figure(figsize=(8, 4))
genre_watch.plot(kind="bar", color="skyblue")
plt.title("Most Watched Genres (Total Watch Hours)")
plt.ylabel("Watch Hours")
plt.xlabel("Genre")
plt.tight_layout()
plt.show()
```

```python
genre_ratings = df.groupby("Genre")["Rating"].mean().sort_values()
plt.figure(figsize=(8, 4))
genre_ratings.plot(kind="barh", color="coral")
plt.title("Average Ratings per Genre")
plt.xlabel("Average Rating")
plt.tight_layout()
plt.show()
```

```python
bingers = df[df["Watch_Hours"] > 15]
binge_genre = bingers["Genre"].value_counts()
plt.figure(figsize=(6, 4))
binge_genre.plot(kind='pie', autopct='%1.1f%%', startangle=90, colors=sns.color_palette("pastel"))
plt.title("Preferred Genres of Binge Watchers")
plt.ylabel("")
plt.tight_layout()
plt.show()
```

```python
print("Behavioral Insights Summary:\n")

top_genre = genre_watch.idxmax()
print(f"Viewers spend the most time on: {top_genre}")

binge_fav = binge_genre.idxmax()
print(f"Binge watchers prefer: {binge_fav} content (>15 hours)")

best_rated_genre = genre_ratings.idxmax()
print(f"Highest-rated genre is: {best_rated_genre} (avg rating: {genre_ratings.max():.2f})")
```

```python
print("\nRecommendation Strategy:")
print(f"- Promote more {best_rated_genre} and {binge_fav} content")
print(f"- Prioritize {top_genre} in homepage carousels")
```
