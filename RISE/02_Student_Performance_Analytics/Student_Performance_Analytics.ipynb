```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("student_performance.csv")
```

```python
print("Summary Statistics:\n")
print(df.describe())
```

```python
avg_marks = df["Marks"].mean()
avg_attendance = df["Attendance (%)"].mean()
avg_logins = df["Login_Frequency"].mean()
at_risk = df[(df["Marks"] < 50) | (df["Attendance (%)"] < 60)]
at_risk_count = len(at_risk)

print(f"ACADEMIC DASHBOARD METRICS")
print(f"Average Marks           : {avg_marks:.2f}")
print(f"Average Attendance (%)  : {avg_attendance:.2f}")
print(f"Average Login Frequency : {avg_logins:.2f}")
print(f"Total At-Risk Students  : {at_risk_count} / {len(df)}\n")
```

```python
plt.figure(figsize=(8, 4))
top10 = df.sort_values(by="Marks", ascending=False).head(10)
sns.barplot(x="Student_ID", y="Marks", data=top10, palette="Blues_d")
plt.title("Top 10 Students by Marks")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
at_risk = df[(df['Marks'] < 50) | (df['Attendance (%)'] < 60)]
print("\nAt-Risk Students (Marks < 50 or Attendance < 60%):")
print(at_risk)
```

```python
plt.figure(figsize=(8, 4))
sns.histplot(df["Attendance (%)"], bins=10, kde=True, color='salmon')
plt.axvline(60, color='red', linestyle='--', label='Risk Threshold (60%)')
plt.title("Distribution of Attendance")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(6, 4))
sns.heatmap(df.drop("Student_ID", axis=1).corr(), annot=True, cmap="YlGnBu")
plt.title("Performance Factor Correlations")
plt.tight_layout()
plt.show()
```
